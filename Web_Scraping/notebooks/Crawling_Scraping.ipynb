```python
compteur = 1
for i in range(1, 4):
  url = f"https://www.chucknorrisfacts.fr/facts/top/{i}"
  site = requests.get(url).text
  soup = BeautifulSoup(site, "html.parser")
  print(f"\n Page {compteur} ------------------------------------------------- \n")
  for fact in soup.find_all("p", class_="card-text"):
    print(fact.text)
  compteur = compteur + 1
```

## Version propre

Les cellules ci-dessus servent à explorer le site. Le code final (parsing, politesse réseau, reprise, vérification de `robots.txt`) est dans [`src/webscraping/chucknorris.py`](../src/webscraping/chucknorris.py) et testé dans `tests/`.

```python
import sys

sys.path.append("../src")  # le code réutilisable est dans src/webscraping

from webscraping.chucknorris import crawl, page_url, scrape_page
from webscraping.storage import save_dataset
```

```python
# Démo sur 3 pages en suivant le lien ">>"
# crawl() vérifie robots.txt, envoie un User-Agent, attend `delay` secondes entre les pages
# et réessaie automatiquement les erreurs temporaires (429, 5xx)
df_demo = crawl(max_pages=3, delay=1)
df_demo.head()
```

```python
# Crawl complet : ~528 pages, environ 9 minutes avec delay = 1
# Le fichier de reprise est écrit toutes les 10 pages : si le crawl est interrompu,
# relancer cette cellule reprend après la dernière page enregistrée.
df_facts = crawl(checkpoint="../data/chuck_norris_facts_partiel.csv", save_every=10)
save_dataset(df_facts, "chuck_norris_facts", "../data")
print(f"{len(df_facts)} faits enregistrés")
```

```python
# Premier fait de la dernière page (528)
facts, _ = scrape_page(page_url(528))
facts[0]
```
